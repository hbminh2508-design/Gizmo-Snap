### Bài W7A1
Viết hàm tìm kiếm nhị phân trên một danh sách đã được sắp xếp tăng dần. Trả về chỉ số của phần tử hoặc `-1` nếu không tìm thấy.
**Mẫu hàm:** `def binary_search(arr, left, right, target):`

```python
# Nhị phân không thể chạy theo một danh sách lộn xộn (Phải dùng sorted())

def binary_search(arr, left, right, target):
  if left > right:
    return -1

  mid = (left + right)//2

  if arr[mid] == target:
    return mid

  elif arr[mid] > target:
    return binary_search(arr, left, mid - 1, target)

  else:
    return binary_search(arr, mid + 1, right, target)

arr = sorted([int(x) for x in input().split()])
target = int(input())

result = binary_search (arr, 0, len(arr) - 1, target)
print(result)
```

### Bài W7A2
Cho danh sách `arr` và một số nguyên `x`, đếm xem `x` xuất hiện bao nhiêu lần trong danh sách.
**Mẫu hàm:** `def count_occurrences(arr, x) -> int:`

### Bài W7A3
Viết hàm thực hiện thuật toán Sắp xếp nổi bọt (Bubble Sort) và Sắp xếp chọn (Selection Sort) để sắp xếp danh sách theo thứ tự tăng dần.

```python
def Bubble_Sort(arr):
  n = len(arr)

  for i in range(n):
    swapped = False
    for j in range(0, n - i - 1): # -> n - i - 1 vì do sau mỗi lần đổi chỗ thì số lớn nhất bị đưa ra sau rồi
      if arr[j] > arr[j + 1]:
        arr[j], arr[j + 1] = arr[j + 1], arr[j]
        swapped = True

    if not swapped:
      break

  return arr

input_data = input().split()
arr = [int(x) for x in input_data]

gia_tri_sap_xep = Bubble_Sort(arr)
print(" ".join(map(str, gia_tri_sap_xep)))
```

### Bài W7A4
Viết hàm tìm phần tử có số lần xuất hiện nhiều nhất trong danh sách. Nếu có nhiều phần tử cùng số lần, trả về phần tử xuất hiện trước.

```python
def max_in_array(arr):
  if not arr:
    return None
  if len(arr) == 1:
    return arr[0]

  dem = {}
  for x in arr:
    if x in dem:
      dem[x] += 1
    else:
      dem[x] = 1

  max_freq = 0
  for val in dem.values():
    if val > max_freq:
      max_freq = val

  for x in arr:
    if dem[x] == max_freq:
      return x, max_freq

lst = list(map(int, input().split()))
val, count = max_in_array(lst)
print(f"{val} xuất hiện nhiều nhất, sớm nhất, {count} lần")
```

### Bài W7A5
Cho danh sách các số nguyên và một số `X`, đếm tất cả các cặp `(a, b)` sao cho `a + b = X`.

```python
def cac_cap_so_bang_x(arr, x):
  dem = 0
  for i in range(len(arr)):
    for j in range(i + 1, len(arr)):
      if arr[i] + arr[j] == x:
        dem += 1
  return dem

lst = list(map(int, input().split()))
X = int(input())

print(cac_cap_so_bang_x(lst, X))
```

### Bài W7A7
Cho một danh sách các số nguyên `nums`, hãy tìm độ dài của dãy con tăng nghiêm ngặt dài nhất (không nhất thiết phải liền kề).

```python
def lis(arr):
  if not arr:
    return 0

  n = len(arr)
  dp = [1] * n

  for i in range(1, n):
    for j in range(i):
      if arr[i] > arr[j]:
        if dp[i] < dp[j] + 1:
          dp[i] = dp[j] + 1

  max_length = 0
  for length in dp:
    if length > max_length:
      max_length = length
  return max_length

nums = input()
clean_nums = [int(x) for x in nums.replace('[', ' ').replace(']', ' ').replace(',', ' ').split()]
print(lis(clean_nums))
```

### Bài W7A8
Bạn được cung cấp một danh sách các khoảng thời gian `interval = [[l1, r1], [l2, r2], ...]` và một danh sách các truy vấn `queries = [q1, q2, ...]`.
Với mỗi truy vấn `q`, hãy tìm độ dài nhỏ nhất của khoảng thời gian `[li, ri]` sao cho `li <= q <= ri`. Nếu không tồn tại khoảng thời gian nào như vậy, hãy trả về `-1` cho truy vấn đó.
**Gợi ý:** Giải trong thời gian `n < O(m.n)`, hint: binary search in the queries list.

```python
def get_len(inter):
    return inter[1] - inter[0] + 1

def minInterval(intervals, queries):
    intervals.sort(key=get_len)

    q_list = []
    for i in range(len(queries)):
        q_list.append([queries[i], i])

    q_list.sort()

    ans = []
    for i in range(len(queries)):
        ans.append(-1)

    for inter in intervals:
        l = inter[0]
        r = inter[1]
        dodai = r - l + 1

        dau = 0
        cuoi = len(q_list) - 1
        vitri = len(q_list)

        while dau <= cuoi:
            giua = (dau + cuoi) // 2
            if q_list[giua][0] >= l:
                vitri = giua
                cuoi = giua - 1
            else:
                dau = giua + 1

        for j in range(vitri, len(q_list)):
            giatri_q = q_list[j][0]
            chi_so_goc = q_list[j][1]

            if giatri_q > r:
                break

            if ans[chi_so_goc] == -1:
                ans[chi_so_goc] = dodai

    return ans

n = int(input())
intervals = []
for i in range(n):
    nhap = input().split()
    l = int(nhap[0])
    r = int(nhap[1])
    intervals.append([l, r])

nhap_q = input().split()
queries = []
for x in nhap_q:
    queries.append(int(x))

ket_qua = minInterval(intervals, queries)

chuoi_kq = ""
for kq in ket_qua:
    chuoi_kq += str(kq) + " "
print(chuoi_kq.strip())
```

### Bài W7A9
Cho một danh sách các số `nums`, với mỗi phần tử `nums[i]`, hãy đếm xem có bao nhiêu phần tử bên phải nhỏ hơn `nums[i]`. Trả về danh sách các số đếm.
**Gợi ý:** Efficient solution: using heap or binary search tree.

```python
def insert_node(tree, node_id, val, total_smaller, next_id):
    if node_id is None:
        tree[next_id[0]] = {
            'val': val,
            'left_size': 0,
            'count': 1,
            'left': None,
            'right': None
        }
        current_id = next_id[0]
        next_id[0] += 1
        return current_id, total_smaller

    node = tree[node_id]

    if val == node['val']:
        node['count'] += 1
        return node_id, total_smaller + node['left_size']
    elif val < node['val']:
        node['left_size'] += 1
        left_child_id, total_smaller = insert_node(tree, node['left'], val, total_smaller, next_id)
        node['left'] = left_child_id
    else:
        total_smaller += node['left_size'] + node['count']
        right_child_id, total_smaller = insert_node(tree, node['right'], val, total_smaller, next_id)
        node['right'] = right_child_id

    return node_id, total_smaller

def countSmaller(nums):
    if not nums:
        return []

    result = [0] * len(nums)
    tree = {}
    root_id = None
    next_id = [0]

    for i in range(len(nums) - 1, -1, -1):
        root_id, total_smaller = insert_node(tree, root_id, nums[i], 0, next_id)
        result[i] = total_smaller

    return result

nums = [5, 2, 6, 1]
print(countSmaller(nums))
```

### Bài W7A10
Cho một danh sách các số nguyên `nums` và một số nguyên `m`, hãy tìm tổng mảng con lớn nhất theo modulo `m`. Tức là, với bất kỳ mảng con `nums[i:j]` nào, hãy tính `sum(nums[i:j]) % m` và trả về kết quả lớn nhất có thể.
**Gợi ý:** Duy trì tổng tiền tố (prefix sum) mod `m` và sử dụng tìm kiếm nhị phân để tìm tổng tiền tố nhỏ nhất chỉ lớn hơn tổng hiện tại một chút (để tối đa hóa hiệu số bao quanh).

```python
def binary_search_closest_greater(sorted_list, target):
    low = 0
    high = len(sorted_list) - 1
    ans = -1

    while low <= high:
        mid = (low + high) // 2
        if sorted_list[mid] > target:
            ans = mid
            high = mid - 1
        else:
            low = mid + 1

    return ans

def insert_in_order(sorted_list, val):
    low = 0
    high = len(sorted_list) - 1
    pos = len(sorted_list)

    while low <= high:
        mid = (low + high) // 2
        if sorted_list[mid] >= val:
            pos = mid
            high = mid - 1
        else:
            low = mid + 1

    sorted_list.insert(pos, val)

def maxSubarrayMod(nums, m):
    max_mod_sum = 0
    current_prefix = 0
    seen_prefixes = [0]

    for num in nums:
        current_prefix = (current_prefix + num) % m
        max_mod_sum = max(max_mod_sum, current_prefix)

        idx = binary_search_closest_greater(seen_prefixes, current_prefix)
        if idx != -1:
            max_mod_sum = max(max_mod_sum, (current_prefix - seen_prefixes[idx] + m) % m)

        insert_in_order(seen_prefixes, current_prefix)

    return max_mod_sum

nums = [3, 3, 9, 9, 5]
m = 7
print(maxSubarrayMod(nums, m))
```

### Bài W7A11
Các trình duyệt web thường có một chức năng tìm kiếm một từ trong một trang web... Hãy viết một chương trình để xác định xem có bao nhiêu từ xuất hiện trong một câu.
**Đầu vào:** Một chuỗi kí tự trên một dòng. Từ khóa cần tìm ở dòng tiếp theo.
**Đầu ra:** Số lần xuất hiện của từ tìm kiếm.

```python
def count_word_occurrences(sentence, keyword):
    punctuation = ".,!?;:\"'()[]{}-_*/\\"
    for char in punctuation:
        sentence = sentence.replace(char, " ")

    words = sentence.lower().split()
    target = keyword.lower().strip()

    count = 0
    for word in words:
        if word == target:
            count += 1

    return count

input_sentence = input()
input_keyword = input()

print(count_word_occurrences(input_sentence, input_keyword))
```

### Bài W7A12
Sắp xếp các phương án theo một trật tự nhất định được quy định trong câu hỏi. Hãy bắt đầu bằng việc sắp xếp các cụm từ dưới đây theo thứ tự xuất hiện trong từ điển.
**Đầu vào:** 4 cụm từ xuất hiện trên bốn dòng tương ứng với các phương án: A, B, C và D.
**Đầu ra:** Thứ tự đúng của các phương án, cách nhau bởi dấu cách.

```python
def sort_options():
    lines = []
    labels = ['A', 'B', 'C', 'D']

    for label in labels:
        line = input()
        lines.append((label, line))

    for i in range(len(lines)):
        for j in range(0, len(lines) - i - 1):
            if lines[j][1] > lines[j+1][1]:
                lines[j], lines[j+1] = lines[j+1], lines[j]

    result = []
    for item in lines:
        result.append(item[0])

    print(" ".join(result))

sort_options()
```
